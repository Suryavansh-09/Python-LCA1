```python

import pandas as pd

# Creating a sample dataset
data = {
    'Name': ['Amit', 'Rahul', 'Priya', 'Sneha', 'Amit'],
    'Marks': [85, 90, None, 75, 85],
    'City': ['Pune', 'Mumbai', 'Pune', None, 'Pune']
}

# Creating a DataFrame
df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
print(df)

# Display first five rows
print("\nFirst Five Rows:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display statistical description
print("\nStatistical Description:")
print(df.describe())

# Checking missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Filling missing values
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['City'] = df['City'].fillna('Unknown')

print("\nDataset After Filling Missing Values:")
print(df)

# Checking duplicate records
print("\nDuplicate Records:")
print(df.duplicated())

# Removing duplicate records
df = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df)

# Selecting data using loc[]
print("\nSelected Data:")
print(df.loc[:, ['Name', 'Marks']])

# Filtering records
print("\nStudents Having Marks Greater Than 80:")
print(df.loc[df['Marks'] > 80])

# Final dataset
print("\nFinal Preprocessed Dataset:")
print(df)
```
